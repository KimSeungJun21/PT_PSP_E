```python
import json
import os,sys
import random
```

```python
label_file_path = '/home/kimseungjun/datasets/My_PT_data/PT_data/train/label_stats.json'
with open(label_file_path,'r') as f:
    data = json.load(f)
```

```python
# 1. 라벨별로 파일명을 분류합니다.
neg_files = [fn for fn, l in data['labels'].items() if l == 0]
pos_files = [fn for fn, l in data['labels'].items() if l == 1]

# 2. 더 적은 쪽의 개수를 확인합니다.
min_count = min(len(neg_files), len(pos_files))

# 3. 양쪽에서 동일한 개수만큼 무작위로 추출합니다. (랜덤 셔플 권장)
selected_neg = random.sample(neg_files, min_count)
selected_pos = random.sample(pos_files, min_count)

# 4. 새로운 딕셔너리로 합칩니다.
new_label = {}
for fn in selected_neg:
    new_label[fn] = 0
for fn in selected_pos:
    new_label[fn] = 1

print(f"Negative: {len(selected_neg)}, Positive: {len(selected_pos)}")

```

```python
import os
import shutil
import json

# 1. 경로 설정
source_dir = '/home/kimseungjun/datasets/My_PT_data/PT_data/train' # 원본 이미지 폴더
target_dir = '/home/kimseungjun/datasets/My_PT_data/PT_data/train_balanced' # 복사될 새 폴더
if not os.path.exists(target_dir):
    os.mkdir(target_dir)
new_label_save_path = '/home/kimseungjun/datasets/My_PT_data/PT_data/train_balanced/label_balanced.json'


print(f"복사를 시작합니다. 총 파일 수: {len(new_label)}")

# 3. 파일 복사 루프
copy_count = 0
for file_name in new_label.keys():
    src_path = os.path.join(source_dir, file_name)
    dst_path = os.path.join(target_dir, file_name)
    
    # 원본 파일이 존재하는지 확인 후 복사
    if os.path.exists(src_path):
        shutil.copy2(src_path, dst_path) # copy2는 메타데이터(수정시간 등)까지 복사합니다
        copy_count += 1
    else:
        print(f"경고: 파일을 찾을 수 없습니다 -> {src_path}")

# 4. 균형 잡힌 새 라벨 파일도 저장 (나중에 학습 시 사용)
with open(new_label_save_path, 'w') as f:
    json.dump(new_label, f, indent=4)

print(f"--- 작업 완료 ---")
print(f"복사된 파일 개수: {copy_count}")
print(f"새 라벨 파일 저장 위치: {new_label_save_path}")
```
